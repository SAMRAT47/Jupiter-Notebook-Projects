```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import joblib
import pickle
```

```python
# Load dataset
df = pd.read_csv('india.csv')
```

```python
# Create a copy of the DataFrame
df_copy = df.copy()
```

```python
df_copy.columns
```

```python
# Convert 'date' column to datetime format
df_copy['date'] = pd.to_datetime(df_copy['date'], dayfirst=True)
```

```python
#Removing extra whitespace from column name
df.columns=[col.strip() for col in df.columns]
print(df.head())  # View first few rows
```

```python
print(df_copy.info())  # View data types and missing values
```

```python
print(df_copy.describe())
```

```python
#Handle missing values
numeric_cols = df_copy.select_dtypes(include=['float64']).columns
df_copy[numeric_cols] = df_copy[numeric_cols].fillna(df_copy[numeric_cols].mean())
```

```python
#Visualize confirmed cases over time
plt.figure(figsize=(10,6))
plt.plot(df_copy['date'], df_copy['confirmed'])
plt.title('Confirmed Cases Over Time')
plt.xlabel('Date')
plt.ylabel('Confirmed Cases')
plt.show()
```

```python
# Test stationarity using Augmented Dickey-Fuller test
df_copy.set_index('date', inplace=True)
result = adfuller(df_copy['confirmed'])
print('ADF Statistic:', result[0])
print('p-value:', result[1])
df_copy.reset_index(inplace=True)

# Decompose time series into trend, seasonal, and residual components
df_copy.set_index('date', inplace=True)
df_copy.index = pd.to_datetime(df_copy.index)
df_copy.index.freq = 'D'  
decomposition = seasonal_decompose(df_copy['confirmed'], model='additive')
trend = decomposition.trend
seasonal = decomposition.seasonal
residual = decomposition.resid
df_copy.reset_index(inplace=True)


# Create a rainbow color map
cmap = plt.cm.rainbow
colors = [cmap(i) for i in np.linspace(0, 1, 4)]  # 4 subplots

plt.figure(figsize=(10,8))
plt.subplot(411)
plt.plot(df_copy['confirmed'], label='Original',color='b')
plt.legend(loc='best')
plt.subplot(412)
plt.plot(trend, label='Trend',color='g')
plt.legend(loc='best')
plt.subplot(413)
plt.plot(seasonal, label='Seasonality',color='c')
plt.legend(loc='best')
plt.subplot(414)
plt.plot(residual, label='Residuals',color='r')
plt.legend(loc='best')
plt.tight_layout()
plt.show()
```

```python
# Split data into training and testing sets
train, test = train_test_split(df_copy, test_size=0.2, random_state=42, shuffle=False)

# Create and fit ARIMA model
train_copy = train.copy()
train_copy.set_index('date', inplace=True)
train_copy.index = pd.to_datetime(train_copy.index)
train_copy.index.freq = 'D'
model = ARIMA(train_copy['confirmed'], order=(5,1,0), freq='D')
model_fit = model.fit()


# Reset index
train_copy.reset_index(inplace=True)

# Plot residuals
residuals = pd.DataFrame(model_fit.resid)
plt.figure(figsize=(10,6))
plt.plot(residuals)
plt.title('Residuals')
plt.show()
```

```python
# Evaluate ARIMA model
forecast = model_fit.forecast(steps=len(test))
arima_mse = mean_squared_error(test['confirmed'], forecast)
arima_r2 = r2_score(test['confirmed'], forecast)
arima_mae = mean_absolute_error(test['confirmed'], forecast)
print('ARIMA MSE:', arima_mse)
print('ARIMA R2:', arima_r2)
print('ARIMA MAE:', arima_mae)
```

```python
# Convert datetime to numeric values
X = pd.to_numeric(df_copy['date'].apply(lambda x: x.timestamp()))

# Reshape X to 2D array
X = X.values.reshape(-1, 1)

# Create and fit Linear Regression model
model_lr = LinearRegression()
model_lr.fit(X, df_copy['confirmed'])

# Plot predicted values
plt.figure(figsize=(10,6))
plt.plot(df_copy['date'], df_copy['confirmed'], label='Actual')
plt.plot(df_copy['date'], model_lr.predict(X), label='Predicted')
plt.title('Actual vs Predicted')
plt.xlabel('Date')
plt.ylabel('Confirmed Cases')
plt.legend()
plt.show()
```

```python
# Evaluate Linear Regression model
lr_forecast = model_lr.predict(X)
lr_mse = mean_squared_error(df_copy['confirmed'], lr_forecast)
lr_r2 = r2_score(df_copy['confirmed'], lr_forecast)
lr_mae = mean_absolute_error(df_copy['confirmed'], lr_forecast)
print('Linear Regression MSE:', lr_mse)
print('Linear Regression R2:', lr_r2)
print('Linear Regression MAE:', lr_mae)
```
