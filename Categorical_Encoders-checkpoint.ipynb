# <mark style="color:Purple"> Categorical encoders

**`Categorical data`** describes the characteristics of an entity.

__Nominal__ – These are variables which are not related to each other in any order such as colour (black, blue, green).

__Ordinal__ – These are variables where a certain order can be found between them such as student grades (A, B, C, D, Fail).

The common examples and values of categorical data are –

- Gender: Male, Female, Others
- Education qualification: Primary, High school, Undergraduate, Master’s or PhD
- City: Bangalore, Mumbai, Delhi, or Chennai, and so on.

Usually, `categorical data` are **`non-numeric`** in nature and are text. It is, however, possible that categorical data is denoted by numbers e.g.  Gender can be denoted by numbers like Male = 1, Female = 2, Others = 3, and so on but it does not have any mathematical significance.

__Types of encoders__

- **OrdinalEncoder** - will assign a number to categorical X variables.
- **LabelEncoder**   - will assign a number to categorical Y variables(target variables).
- **OneHotEncoder**  - will create a separate "column" for each category of each X variable.
- **LocalBinarizer**  - will create a separate "column" for each category of each X variable.

<code> **Note:**
Before we look into how these encoders work, we can perform this encoding using pandas when you have very less categories.

```python
#import libraries
import numpy as np
import pandas as pd
import random
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,OrdinalEncoder
```

```python
# syntax: pd.get_dummies(df,['var','var1',......'varn'])
```

## Ordinal Encoder

```python
# encode variables
planets_choices = ["Earth","Mars","Jupiter","Saturn","Venus"]
x_values_train = np.array([[_] for _ in planets_choices])
x_values_test = np.array([["Earth"], ["Mars"],["Jupiter"],["Saturn"],["Venus"]])
```

```python
ord_enc = OrdinalEncoder()
ord_enc.fit(x_values_train)
ord_enc.transform(x_values_test)
```

## Label Encoder

```python
l_enc = LabelEncoder()
l_enc.fit(x_values_test)
pd.DataFrame(l_enc.transform(x_values_test))
```

## OneHot Encoder

```python
ohe_enc = OneHotEncoder()
ohe_enc.fit(x_values_train)
pd.DataFrame(ohe_enc.transform(x_values_test).toarray(), columns=ohe_enc.categories_)
```

## Label Binarizer

```python
from sklearn.preprocessing import LabelBinarizer
```

```python
# label binarizer used to label specific variables, 
lb = LabelBinarizer()
lb.fit([1, 2, 6, 4, 2])
LabelBinarizer()
print(lb.classes_)
lb.transform([1, 6])
```

```python
lb_style = LabelBinarizer()
lb_results = lb_style.fit_transform(x_values_train)
pd.DataFrame(lb_results)
```
