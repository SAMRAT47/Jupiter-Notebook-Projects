##  <mark><span  style= "font-family: Times New Roman">Import necessary libraries</span> </mark> <a class="anchor" id="Import_necessary_libraries"></a>

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

##  <mark><span  style= "font-family: Times New Roman">View the data</span> </mark> <a class="anchor" id="Import_necessary_libraries"></a>

```python
data={'Batsman 1 Score':[56,19,0,68,16,67,85,101,78],'Batsman 2 Score':[60,58,38,71,69,66,68,41,67],'Temperature':[31,33,30,35,34,15,10,7,11]}
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```


##  <mark><span  style= "font-family: Times New Roman">Data preprocessing</span> </mark> <a class="anchor" id="Import_necessary_libraries"></a>

<div class="warning" style='padding:0.1em; background-color:#E9D8FD; color:#69337A'>
<span>
<p style='margin-top:1em; text-align:left'>
</p>
<p style='margin-left:1em;'>
</p><br>
   
  1. Third match score of Batsman-1 was mistakenly noted as 0. Kindly replace the value with the mean of earlier matches as it is not known.Round the calculated mean to the nearest integer. <br>    

2. Create a new feature in the dataframe naming 'weather' <br>
    
   a. Column should have 'Warm' if the temperature on the match day is greater than the mean of the temperatures.  <br>
   b. Column should have 'Cool' if the temperature on the match day is lesser than the mean of the temperatures.  <br>
</span>
</div>


```python
#Calculate mean of Batsman 1 Scrore and round-off to nearest integer
mean_batsman1=round(df['Batsman 1 Score'].mean())
print(mean_batsman1)
```

```python
#Replace the incorrect value(0) with the mean
df['Batsman 1 Score']=df['Batsman 1 Score'].replace(0,mean_batsman1)
print(df)
```

```python
#Calculate mean of Temperature
mean_temp=df['Temperature'].mean()
print(mean_temp)
```

```python
#Creating new feature and applying categories based on condition
df['Weather']=np.where(df['Temperature']>mean_temp,'Warm','Cool')
print(df)
```

##  <mark><span  style= "font-family: Times New Roman">Perform EDA</span> </mark> <a class="anchor" id="Import_necessary_libraries"></a>

### Visualize the evidences 

```python
'''
Filtering data for Cool weather as England is generally considered a cool weather country
and test series is happening during winter. winter.
'''
cool_weather_df = df[df['Weather'] == 'Cool']
print(cool_weather_df)
```

```python
#EDA
print("Data Shape:",cool_weather_df.shape)
print("Data Description:\n",cool_weather_df.describe())
```

```python
plt.figure(figsize=(10,6))

sns.histplot(cool_weather_df['Batsman 1 Score'], color='blue', label='Batsman 1',alpha=0.5)
sns.histplot(cool_weather_df['Batsman 2 Score'], color='red', label='Batsman 2',alpha=0.5)
plt.title('Batsmen Score Distribution', fontsize=12, ha='center')
plt.xlabel('Score', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.legend(['Batsman 1', 'Batsman 2'], loc='upper right')
```

```python
# Comparison of Batsmen Avearage Score in Cool Weather
plt.figure(figsize=(10, 10))

plt.subplot(2, 2, 1)
sns.scatterplot(x='Temperature', y='Batsman 1 Score', data=cool_weather_df, label='Batsman 1', color='blue',alpha=0.8)
sns.scatterplot(x='Temperature', y='Batsman 2 Score', data=cool_weather_df, label='Batsman 2', color='red',alpha=0.8)
plt.title('Temperature vs Batsmen Score', fontsize=12, ha='center')
plt.xlabel('Temperature', fontsize=10)
plt.ylabel('Score', fontsize=10)

plt.subplot(2, 2, 2)
sns.barplot(x=['Batsman 1', 'Batsman 2'], y=[cool_weather_df['Batsman 1 Score'].mean(), cool_weather_df['Batsman 2 Score'].mean()],hue=['Batsman 1', 'Batsman 2'],palette=['blue','red'],alpha=0.8)
plt.title('Batsmen Average Score in Cool Weather')
plt.xlabel('Batsman')
plt.ylabel('Average Score')
plt.show()
```

```python
Average_Score_of_Batsmen_in_Cool_weather={'Batsman_1':cool_weather_df['Batsman 1 Score'].mean(),'Batsman_2':cool_weather_df['Batsman 2 Score'].mean()}
print(Average_Score_of_Batsmen_in_Cool_weather)
```
